### 4.可以开启你的小组项目的第一个小小任务啦！就是对一小段有关“功勋科学家”的文本进行分词处理。请对以上第三段文本进行分词，并评估分词效果（哪些地方分的好，哪些分的不好）。

```python
# 4. 对第三段"功勋科学家"文本做分词，并评估分词效果
import jieba

text_huang = ('黄旭华，1926年3月12日出生于广东省汕尾市，原籍广东省揭阳市。'
              '1949年毕业于上海交通大学。历任北京海军核潜艇研究室副总工程师、'
              '中船重工集团公司核潜艇总体研究设计所研究员、名誉所长。'
              '1994年当选为中国工程院院士。')

seg_huang_default = jieba.cut(text_huang)
print('【默认分词结果】')
print('/'.join(seg_huang_default))
```

```python
# 载入用户词典后再分词，观察机构名、头衔能否被整体切出
jieba.load_userdict('dict.txt')

seg_huang_dict = jieba.cut(text_huang)
print('【加载用户词典后】')
print('/'.join(seg_huang_dict))
```

```python
# 叠加停用词过滤，去掉标点与虚词
stopwords = [line.strip() for line in open('stop_words.txt', 'r', encoding='utf-8').readlines()]

final_huang = ''
for seg in jieba.cut(text_huang):
    if seg not in stopwords and seg.strip() != '':
        final_huang += seg + '/'

print('【去停用词后】')
print(final_huang)
```

### 分词效果评估

#### 一、分得好的地方

| 类型 | 切分结果 | 说明 |
| --- | --- | --- |
| 人名 | 黄旭华 | 姓名完整，没有被拆成"黄/旭/华" |
| 地名 | 广东省 / 汕尾市 / 揭阳市 | 省、市级行政区完整 |
| 高校名 | 上海交通大学 | 校名完整，没有被切成"上海/交通/大学" |
| 时间 | 1926年 / 3月 / 1949年 / 1994年 | 年月切分符合中文表达习惯 |
| 职称 | 研究员 / 所长 / 院士 | 职务后缀识别正确 |
| 动词短语 | 毕业 / 当选 / 历任 | 动宾搭配切分合理 |

#### 二、分得不好的地方

1. **"12日出"** —— 正确结果应是 `12日 / 出生`。jieba 把"日出"当成一个词，属于典型的未登录词歧义错误。这个错误会直接污染后续词频统计，凭空多出一个"日出"。
2. **"中 / 船 / 重工 / 集团公司"** —— 机构名被切成四段。默认词典里没有"中船重工集团公司"这个专有名词，**加载用户词典后修复为整体**，这是词典最直接的收益。
3. **"副 / 总工程师"** —— "副总工程师"是完整职务，却被拆成"副 + 总工程师"。**即使加载了用户词典也没有改善**，说明 dict.txt 里缺少这个条目，词表需要持续补充。
4. **"总体 / 研究 / 设计所"** —— "核潜艇总体研究设计所"这类超长机构名，词典几乎无法穷举，仍是碎片化输出。这类问题靠词典解决不了，要上专业 NER 模型。
5. **"中国工程院 / 院士" → "中国工程院院士"** —— 加入词典后变成整体切分。但要注意：这一步未必总是更好，如果后续想统计"院士"这个身份出现了几次，整体切分反而会让"院士"统计不到。**分词粒度应当服务于下游任务。**

#### 三、结论

- jieba 在通用词汇上表现稳定，**主要失分点集中在专有名词**：机构名、职务名、学科术语。这也正是后面要做 NER（命名实体识别）的原因——规则和词典都很难覆盖全。
- 改进思路有三条：
  1. **补充领域用户词典**：把本领域高频机构名、术语加进 dict.txt（本例中"中船重工集团公司""中国工程院院士"就是这么修好的）；
  2. **加词性过滤**：用 `jieba.posseg` 只保留名词类（n、nr、ns、nt），比单纯用停用词表更彻底；
  3. **换用 NER 模型**：对机构名、职务这类长实体，用 BERT 类序列标注模型效果远好于词典。
- 停用词过滤能清掉标点和"的、了、是"等虚词，但本例的 stop_words.txt 只有 7 个词，"于""为""历任"这类仍会残留。真实项目要换用完整的停用词表（如哈工大停用词表）。
