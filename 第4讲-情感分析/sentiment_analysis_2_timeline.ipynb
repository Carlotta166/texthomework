情感分析数值可以正确计算。在此基础上，我们需要定义函数，以便批量处理所有的评论信息。

```python
def get_sentiment_cn(text):
    s = SnowNLP(text)
    return s.sentiments  # snownlp是通用的，就没那么准确
```

然后，我们利用Python里面强大的apply语句，来一次性处理所有评论，并且将生成的情感数值在数据框里面单独存为一列，称为sentiment。

```python
df["sentiments"] = df.comments.apply(get_sentiment_cn)  #这行代码很有意思！
```

```python
df.head(100)
```

新的列sentiment已经生成。我们之前介绍过，SnowNLP的结果取值范围在0到1之间，代表了情感分析结果为正面的可能性。通过观察前几条数据，我们发现点评网站上，顾客对这家分店评价总体上还是正面的，而且有的评论是非常积极的。

但是少量数据的观察，可能造成我们结论的偏颇。我们来把所有的情感分析结果数值做一下平均。使用mean()函数即可。

```python
df.sentiments.mean()
```

整体接近0.7，正面。我们再来看看中位数值，使用的函数为median()。

```python
# ggplot 原来是 R 语言里的绘图包，Python 里有对应的移植版 plotnine，用法几乎一样
from plotnine import *

# 上一格已经把情感值存进 sentiments 列，这里按时间排序，保证折线是顺着时间连的
df_plot = df.sort_values("date").reset_index(drop=True)

print("评论条数：", len(df_plot))
print("时间范围：", df_plot.date.min(), "至", df_plot.date.max())
df_plot[["date", "sentiments"]].head()
```

```python
p = (ggplot(df_plot, aes(x="date", y="sentiments"))
     + geom_line(color="#c0392b", linetype="dashed", size=0.7)   # 虚线连接，看趋势
     + geom_point(color="#2c3e50", size=2.4)                     # 实心点，看单条评论
     + labs(x="评论时间",
            y="情感值（0-1，越接近1越正面）",
            title="餐厅评论情感值时间序列")
     + theme_bw()
     + theme(text=element_text(family="Microsoft YaHei", size=11),
             figure_size=(10, 6),
             dpi=140,
             plot_title=element_text(ha="center")))

p.save("timeline_ggplot.png", verbose=False)
p
```

上面这张图跟老师给的散点图比，多了三样东西：

1. **连线**。散点图只能看出"哪一天有低分"，加上虚线之后能看出**走向**——比如某段时间连续走低，说明问题是持续的，不是偶发。
2. **统一的纵轴刻度**。0 到 1 拉满，低分的点落到底部一眼就能认出来，不用去比高度。
3. **中文字体**。`font.family` 设成 Microsoft YaHei，坐标轴和标题才不会变成方块。

有一点值得记下来：图里点的密度并不均匀，有些日子连着好几条，有些空了一大段。这说明数据本身是**不等间隔采样**的，画折线时线段斜率的陡缓会被采样间隔放大或压缩，读趋势时不能只看线的陡峭程度。真要做严格的趋势分析，得先按周或按月聚合，再画。
