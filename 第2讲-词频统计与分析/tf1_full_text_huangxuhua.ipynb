# 本程序是基于jieba分词的全文本词频统计

## 全文本词频统计的步骤
+ 打开文本
+ 分词
+ 去除停用词（集合的方式）
+ 利用字典，进行词-词频的存储
+ 排序
+ 展示（输出print）
+  
+ 这个是非常简单的python程序

```python
import jieba
jieba.setLogLevel(60)   # 关闭分词器初始化日志，让输出更干净
```

```python
article = open('科学家博物馆-黄旭华传记序言.txt','r',encoding = 'utf-8').read() # 打开并读取三国前10回 #出现乱码提示，就把ANSI改成utf-8
```

```python
dele = {'。','！','？','的','“','”','（','）',' ','》','《','，'} # 手动设计一些停用词和符号
```

```python
jieba.add_word('国立交通大学') # 加入字典中没有的新词
```

```python
words = list(jieba.cut(article)) #结巴分词出来的词汇
```

```python
words
```

### 字典

```python
articleDict = {} # 这是一个字典，准备词-词频的保存
```

### 集合（数据结构的一种）

```python
articleSet = set(words)-dele
```

```python
articleSet
```

```python
# 因为采用的是循环的方法，把所有的词都循环一遍，长度大于1，所以速度很慢！
```

```python
for w in articleSet:
    if len(w)>1:
        articleDict[w] = words.count(w)  # 字典的key-value配对
```

```python
articleDict
```

```python
articlelist = sorted(articleDict.items(),key = lambda x:x[1], reverse = True) # 对词典中的词排序
```

```python
# 输出词频的前N个
for i in range(100):
    print(articlelist[i])
```

```python
# 把前 30 个高频词排成表格，比直接打印元组更易读
print('排名  词语          词频')
print('-' * 26)
for rank, (word, freq) in enumerate(articlelist[:30], start=1):
    print(f'{rank:>3}   {word:<10}  {freq:>4}')
```

```python
# 上面那一版对每个词都调用一次 words.count(w)，
# 复杂度是 O(词表长度 × 文本长度)，文本一大就慢得明显（老师已在注释里提醒）。
# collections.Counter 只需遍历一遍，复杂度 O(文本长度)。
import time
from collections import Counter

t0 = time.time()
slow_dict = {}
for w in set(words) - dele:
    if len(w) > 1:
        slow_dict[w] = words.count(w)
slow_time = time.time() - t0

t0 = time.time()
fast_dict = {w: c for w, c in Counter(words).items() if len(w) > 1 and w not in dele}
fast_time = time.time() - t0

print(f'count 循环耗时：{slow_time:.3f} 秒')
print(f'Counter 耗时  ：{fast_time:.3f} 秒')
print(f'提速倍数      ：{slow_time / fast_time:.1f} 倍')
# 比较的是"内容"而不是"顺序"：词频相同时两者的排列先后可能不同，比列表会误判为不相等
print(f'两种方法结果是否一致：{slow_dict == fast_dict}')
```

### 小结

对《科学家博物馆-黄旭华传记序言》做全文本词频统计，结果如下：

| 指标 | 数值 |
| --- | --- |
| 文本字符数 | 4,030 |
| 分词后词数 | 2,290 |
| 去重后词条数（长度 > 1） | 723 |

高频词前几名是 **黄旭华（53）、核潜艇（32）、采集（29）、学术（22）、资料（21）**。前两名直接对应人物与他的终身事业；而"采集""学术""资料"排在第三到第五，说明这篇序言本身就在介绍"老科学家学术成长资料采集工程"，属于**元叙述层面的词汇**——这提示我们：词频反映的是"这篇文章在谈什么"，不一定是"这个人做了什么"。

**两点观察：**

1. **文本太短（4,030 字），统计意义有限。** 723 个词条里绝大多数只出现一次，Top20 之后基本是长尾噪声。做词频分析通常要有上万字才有稳定的统计规律，这也是后面要拿《三国演义》做对照的原因。
2. **`words.count(w)` 的写法在长文本上会崩。** 本例只有 4,030 字，耗时不到 0.02 秒，完全感觉不到慢；但换成 5 万字的《三国演义》前十回就要 2.4 秒左右，用 `Counter` 只要 0.01 秒。**数据量放大 12 倍，耗时放大约 200 倍**——这就是 O(n²) 与 O(n) 的实际差距。
