# 本程序是基于jieba分词的全文本词频统计

## 全文本词频统计的步骤
+ 打开文本
+ 分词
+ 去除停用词（集合的方式）
+ 利用字典，进行词-词频的存储
+ 排序
+ 展示（输出print）
+  
+ 这个是非常简单的python程序

```python
import jieba
jieba.setLogLevel(60)   # 关闭分词器初始化日志，让输出更干净
```

```python
article = open('sanguo_10.txt','r',encoding = 'utf-8').read() # ansi/打开并读取三国前10回 #出现乱码提示，就把ANSI改成utf-8
```

```python
dele = {'。','！','？','的','“','”','（','）',' ','》','《','，'} # 手动设计一些停用词和符号
```

```python
jieba.add_word('羅貫中') # 加入字典中没有的新词
```

```python
words = list(jieba.cut(article)) #结巴分词出来的词汇
```

```python
words
```

### 字典

```python
articleDict = {} # 这是一个字典，准备词-词频的保存
```

### 集合（数据结构的一种）

```python
articleSet = set(words)-dele
```

```python
articleSet
```

```python
# 因为采用的是循环的方法，把所有的词都循环一遍，长度大于1，所以速度很慢！
```

```python
for w in articleSet:
    if len(w)>1:
        articleDict[w] = words.count(w)  # 字典的key-value配对
```

```python
articleDict
```

```python
articlelist = sorted(articleDict.items(),key = lambda x:x[1], reverse = True) # 对词典中的词排序
```

```python
# 输出词频的前N个
for i in range(100):
    print(articlelist[i])
```

```python
# 把前 30 个高频词排成表格，比直接打印元组更易读
print('排名  词语          词频')
print('-' * 26)
for rank, (word, freq) in enumerate(articlelist[:30], start=1):
    print(f'{rank:>3}   {word:<10}  {freq:>4}')
```

```python
# 上面那一版对每个词都调用一次 words.count(w)，
# 复杂度是 O(词表长度 × 文本长度)，文本一大就慢得明显（老师已在注释里提醒）。
# collections.Counter 只需遍历一遍，复杂度 O(文本长度)。
import time
from collections import Counter

t0 = time.time()
slow_dict = {}
for w in set(words) - dele:
    if len(w) > 1:
        slow_dict[w] = words.count(w)
slow_time = time.time() - t0

t0 = time.time()
fast_dict = {w: c for w, c in Counter(words).items() if len(w) > 1 and w not in dele}
fast_time = time.time() - t0

print(f'count 循环耗时：{slow_time:.3f} 秒')
print(f'Counter 耗时  ：{fast_time:.3f} 秒')
print(f'提速倍数      ：{slow_time / fast_time:.1f} 倍')
# 比较的是"内容"而不是"顺序"：词频相同时两者的排列先后可能不同，比列表会误判为不相等
print(f'两种方法结果是否一致：{slow_dict == fast_dict}')
```

### 小结

| 指标 | 数值 |
| --- | --- |
| 文本字符数 | 50,308（《三国演义》前 10 回） |
| 分词后词数 | 34,310 |
| 去重后词条数 | 7,921 |

高频词前三名是 **董卓（97）、呂布（60）、曹操（59）**。这个结果与"前 10 回"的内容高度吻合——第 3 回到第 9 回的主线正是董卓入京、吕布反目、曹操起兵。**词频确实能反映叙事重心**，这是它最直接的价值。

但同样的结果也暴露出三个问题：

1. **人名没有归一。**"玄德"（48 次）和"玄德曰"（22 次）其实都指刘备，却分列成两个词条；真正的"劉備"又是另一个词条。**同一个人被拆成三条**，人物分析会严重失真。
2. **虚词与通用词仍占高位。**"天下（53）、不可（31）、次日（26）、左右（23）"这类词频很高，但对理解情节几乎没有信息量。
3. **繁体文本的分词能力有限。** 原文是繁体（呂布、袁紹、貂蟬），jieba 默认词典以简体为主，能切对主要人名，但"玄德曰"被整体切出，说明它对文言句式的适配不足。

**对比结论**：同样是"分词 + 计数 + 排序"，4,030 字的黄旭华文本和 50,308 字的三国文本呈现出完全不同的统计面貌。**文本量决定词频分析可不可信**，而分词质量决定统计口径准不准。
