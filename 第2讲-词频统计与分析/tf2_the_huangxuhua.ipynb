```python
def make_chinese_plot_ready():
    from matplotlib import rcParams
    # 原代码写的是 'Heiti TC'，那是 macOS 自带字体，Windows 上不存在，
    # 直接跑中文会显示成一排方框。这里改成 Windows 自带字体，并按优先级回退。
    rcParams['font.family'] = ['Microsoft YaHei', 'SimHei', 'SimSun']
    rcParams['axes.unicode_minus'] = False  # 让负号正常显示，不被当成特殊字符
```

```python
# 定义 画图 函数
```

```python
def draw_dict(mydict, figsize=(8, 5)):
    import pandas as pd
    import matplotlib.pyplot as plt
    make_chinese_plot_ready()
    df = pd.DataFrame(list(mydict.items()), columns=['name', 'times'])
    df.set_index('name')['times'].sort_values(ascending=False).plot(kind='bar', figsize=figsize)  # 做好排序
    plt.tight_layout() 
```

## 作业解答

作业要求：把文本替换为《科学家博物馆-黄旭华传记序言.txt》，完成**全文词频统计**与**指定词汇统计**（如"黄旭华""核潜艇"出现次数）。

上面已经用 `data_txt.count(term)` 完成了指定词汇统计。下面补上全文词频统计，再把两种统计口径的结果放在一起对照。

```python
# 一、全文词频统计（沿用 tf1 的思路：分词 → 去停用词 → 计数 → 排序）
import jieba
from collections import Counter

jieba.setLogLevel(60)   # 关闭分词器初始化日志，让输出更干净
dele = {'。', '！', '？', '的', '“', '”', '（', '）', ' ', '》', '《', '，', '\n', '、', '；', '：'}
jieba.add_word('国立交通大学')   # 词典里没有这个词，不加会被切碎

tokens = list(jieba.cut(data_txt))
words = [w for w in tokens if w not in dele and len(w) > 1]
freq = Counter(words).most_common()

print(f'文本字符数：{len(data_txt)}')
print(f'分词后词数：{len(tokens)}')
print(f'去重后词条数：{len(freq)}')
print()
print('全文词频 Top20：')
for i, (w, c) in enumerate(freq[:20], start=1):
    print(f'{i:>2}. {w:<8} {c:>3}')
```

```python
# 二、指定词汇统计（字符串匹配），并画成柱状图
terms2 = ['黄旭华', '核潜艇', '国立交通大学', '院士', '采集']
terms_dict2 = {t: data_txt.count(t) for t in terms2}

print('指定词汇统计结果：')
for t, c in terms_dict2.items():
    print(f'  {t:<12}{c:>4} 次')

draw_dict(terms_dict2, figsize=(7, 4))
```

```python
# 三、两种口径的差异：本次作业最关键的发现
tok = Counter(jieba.cut(data_txt))

print(f"{'词汇':<12}{'字符串匹配':>10}{'分词统计':>10}{'差额':>8}")
print('-' * 44)
for t in terms2:
    c_str = data_txt.count(t)
    c_tok = tok[t]
    print(f'{t:<12}{c_str:>10}{c_tok:>10}{c_str - c_tok:>8}')
```

### 两种口径为什么差了 6 次？

"黄旭华"在原文中出现 **59** 次，但分词统计只有 **53** 次。差的 6 次不是数据出错，而是**分词把词切在了错误的位置**：

| 原文片段 | jieba 的分词结果 | 后果 |
| --- | --- | --- |
| 以黄旭**华为**代表的 | `黄旭` / `华为` / `代表` | "华为"被当成公司名词条，人名被劈开，**丢 4 次** |
| 黄旭**华出**生于 | `黄旭华出` / `生于` | 多切出一个"出"字，**丢 2 次** |

两者相加正好 6 次。这个案例说明三件事：

1. **统计专名（人名、机构名、术语）时，分词结果不一定可靠。** jieba 词典里有"华为"这个高频词条，遇到"黄旭华为…"就会优先切出"华为"，属于典型的**分词歧义**——词频统计的误差源头不在计数，而在分词。
2. **`str.count()` 的字符串匹配反而更适合统计固定词汇。** 它不看上下文，只要字面出现就算，这正是这个 notebook 上半部分用 `data_txt.count(term)` 而不用分词的原因。
3. **但字符串匹配也有代价——它不辨语义。**"核潜艇"和"潜艇"字面不同就会漏统；反过来，"安全"出现在"未发生安全事故"里也会被计入，语义方向完全反了。

**结论**：统计**已知的固定词表**，用字符串匹配（简单、准确）；要**发现有哪些高频词**，才需要分词（能挖出你没预料到的词）。两种方法各有适用场景，不能互相替代。
