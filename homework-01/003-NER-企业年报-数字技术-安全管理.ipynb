```python
import jieba
from collections import Counter
```

```python
# 文本
text = """
落实“企业管理年”主题，加强QEHS三体系建设，
通过自动化、数字化、智能化升级改造加快新一代信息技术与企业生产经营融合，
打造精益制造能力，提升精细化管理水平，助力公司从“制造”升级为“智造”，
从而提高经营效率和效益:通过集团一体化智数管理平台建设与运营，
丰富企业供应链管理、生产工艺控制等管理工具，不断增强生产经营过程数据获取与分析能力，
强化全过程一体化管理，提高自动化、数字化、智能化的供应链管理能力，
为体系安全稳定运行与管理水平提升保驾护航，致力打造安全智能化工厂;
打造助剂互联网技术合作和商务合作平台，构建具有国际竞争力的供应链体系。
"""
```

```python
# 1. 分词处理
words = jieba.lcut(text)
```

```python
words
```

```python
# 2. 定义要统计的特殊词汇
target_words = ['数字化', '智能化', '安全']
```

```python
# 统计词频
word_counts = Counter(words)
```

```python
# 输出特定词汇的词频统计结果
print("特定词汇词频统计结果：")
for word in target_words:
    print(f"'{word}': {word_counts[word]}次")
```

```python
# 输出所有词汇的词频（按频率降序）
print("\n所有词汇词频统计（前20个）：")
for word, count in word_counts.most_common(20):
    print(f"'{word}': {count}次")
```

### 课堂提问的思考

#### 1. 仅仅统计词频，能反映什么？

词频反映的是**话语强度**，也就是企业在年报里"提了多少次"某类议题。它可以作为关注度或重视程度的代理变量：

- 横截面上比较不同企业：年报里"数字化"出现次数越多，说明这家企业越愿意对外强调数字化战略；
- 时间序列上比较同一企业："安全"词频逐年上升，可能对应安全管理投入增加；
- 作为回归模型的自变量，去解释数字化转型绩效、安全投入、企业风险等结果变量。

但必须清醒：**词频只统计"说了几次"，不反映"说了什么方向"**。说"我们大力推进数字化"和说"数字化带来了新的安全隐患"，在词频上是完全一样的贡献。

#### 2. 词频统计本身存在什么问题？去停用词完成了吗？

**没有完成。** 本例直接对原文分词后统计，标点、虚词全部计入，导致 `most_common(20)` 的结果里"的""、"这类词占据高位，真正的主题词被稀释。

除此之外还有几类系统性问题：

| 问题 | 具体表现 | 后果 |
| --- | --- | --- |
| 未去停用词 | 本例现状 | 虚词、标点污染排名 |
| 未做词性过滤 | "通过""提升""加强"等通用动词被计入 | 高频但无区分度 |
| 同义词未归并 | "数字化""数智化""智能化"口径不一 | 主题词频被拆散，低估真实强度 |
| 分词错误 | 如"12日出"这类切错 | 噪声词进入统计 |
| 未处理否定 | "未发生安全事故"也会计入"安全" | 语义方向完全反了 |
| 未做归一化 | 长文本天然词频高 | 跨企业不可比，应除以总词数或取对数 |

#### 3. 仅仅统计词频有什么不足？能确定"数字化"与"安全"的关系吗？如何改进？

**不能确定关系。** 同句共现最多只能说明两者"被一起提到"，无法区分：

- 数字化**提升**了本质安全水平（正向）；
- 数字化本身**带来**新的信息安全风险（负向）；
- 两者只是并列罗列，毫无关联（无关）。

词频是边际分布，丢掉了词序和句法结构，所以**方向、因果、条件统统表达不了**。

改进路径，从轻到重：

1. **共词分析 / 共现网络**：设定窗口（如±10 词）统计共现次数，用 PMI 衡量关联强度。能看出"谁和谁常一起出现"，解决了"有没有关系"，但解决不了"是什么关系"。
2. **依存句法分析**：用 LTP / HanLP 解析句法树，抽取"主谓宾"三元组，能得到"数字化—提升—安全水平"这样的结构化结论。
3. **NER + 关系抽取（RE）**：先识别实体（技术、风险、措施），再用关系分类模型判定实体对之间的关系类型，最终构建**知识图谱**。这正是本讲后面 003、004 两个示例要走的路线。
4. **大语言模型直接抽取**：把文本和抽取要求一起喂给 LLM，让它直接输出三元组 JSON（见 004 的做法）。开发成本最低，效果取决于模型和提示词设计。
5. **加情感/语义方向判定**：在做完共现或句法分析后，再判断关系的正负向，才能回答"数字化到底改善还是恶化了安全"。
