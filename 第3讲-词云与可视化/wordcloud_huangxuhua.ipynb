基于以上设置作业

font_path : string //字体路径，需要展现什么字体就把该字体路径+后缀名写上，如：font_path = '黑体.ttf'

width : int (default=400) //输出的画布宽度，默认为400像素

height : int (default=200) //输出的画布高度，默认为200像素

prefer_horizontal : float (default=0.90) //词语水平方向排版出现的频率，默认 0.9 （所以词语垂直方向排版出现频率为 0.1 ）

mask : nd-array or None (default=None) //如果参数为空，则使用二维遮罩绘制词云。如果 mask 非空，设置的宽高值将被忽略，遮罩形状被 mask 取代。除全白（#FFFFFF）的部分将不会绘制，其余部分会用于绘制词云。如：bg_pic = imread('读取一张图片.png')，背景图片的画布一定要设置为白色（#FFFFFF），然后显示的形状为不是白色的其他颜色。可以用ps工具将自己要显示的形状复制到一个纯白色的画布上再保存，就ok了。

scale : float (default=1) //按照比例进行放大画布，如设置为1.5，则长和宽都是原来画布的1.5倍。

min_font_size : int (default=4) //显示的最小的字体大小

font_step : int (default=1) //字体步长，如果步长大于1，会加快运算但是可能导致结果出现较大的误差。

max_words : number (default=200) //要显示的词的最大个数

stopwords : set of strings or None //设置需要屏蔽的词，如果为空，则使用内置的STOPWORDS

background_color : color value (default=”black”) //背景颜色，如background_color='white',背景颜色为白色。

max_font_size : int or None (default=None) //显示的最大的字体大小

mode : string (default=”RGB”) //当参数为“RGBA”并且background_color不为空时，背景为透明。

relative_scaling : float (default=.5) //词频和字体大小的关联性

color_func : callable, default=None //生成新颜色的函数，如果为空，则使用 self.color_func

regexp : string or None (optional) //使用正则表达式分隔输入的文本

collocations : bool, default=True //是否包括两个词的搭配

colormap : string or matplotlib colormap, default=”viridis” //给每个单词随机分配颜色，若指定color_func，则忽略该方法。



fit_words(frequencies)  //根据词频生成词云
generate(text)  //根据文本生成词云

to_file(filename)   //输出到文件

## 作业：把科学家文本的词云做得更清晰

对应课件第 31 页第 3 项：「采用给的程序，实现一段科学家文本的词云图绘制，越清晰越好。」

上面已经按老师给的流程生成了一版。下面围绕"清晰"这个目标做优化，影响最大的其实是四个参数：

| 目标 | 参数 | 说明 |
| --- | --- | --- |
| 像素够大 | `width`、`height`、`scale` | 最终像素 = 布局尺寸 × `scale` |
| 字够大 | `max_words`、`min_font_size` | 词少一点、字号下限高一点，才不会挤成一片 |
| 读得顺 | `prefer_horizontal` | 设为 1.0 后全部水平排列，比横竖混排易读得多 |
| 配色 | `colormap` | 同色系更专业；彩色能区分词但会分散注意力 |

```python
# 一、先准备一份干净的词频：分词 → 去停用词 → 去掉单字与纯数字
import os
import jieba
from collections import Counter

jieba.setLogLevel(60)
jieba.add_word('国立交通大学')       # 专有名词，词典里没有会被切碎

# 原文件带 BOM，用 utf-8-sig 读取可自动去掉开头的 \ufeff
text_hd = open('功勋科学家黄旭华传记序言文本.txt', encoding='utf-8-sig').read()

stopwords = set(w for w in open('stop_words.txt', encoding='utf-8').read().split('\n') if w)
stopwords |= {'\n', ' ', '（', '）', '“', '”', '、', '；', '：', '《', '》'}

freq = Counter(w for w in jieba.cut(text_hd)
               if len(w) > 1 and w not in stopwords and not w.isdigit())

print(f'文本 {len(text_hd)} 字，有效词条 {len(freq)} 个')
for w, c in freq.most_common(12):
    print(f'  {w:<8}{c:>4}')
```

```python
# 二、字体：中文词云必须指定中文字体，否则画出来全是方框
# 优先用老师提供的 simsun.ttf；目录里没有时回退到系统自带字体
FONT = next((p for p in ['simsun.ttf',
                         r'C:\Windows\Fonts\simsun.ttc',
                         r'C:\Windows\Fonts\msyh.ttc',
                         r'C:\Windows\Fonts\simhei.ttf'] if os.path.exists(p)), None)
print('本次使用的字体：', FONT)
assert FONT, '未找到中文字体，请把 simsun.ttf 放到本文件夹'
```

```python
# 三、版式一：高分辨率矩形版——可读性最好，词频的大小关系一目了然
from wordcloud import WordCloud
import matplotlib.pyplot as plt

wc_rect = WordCloud(
    font_path=FONT,
    width=1600, height=1000, scale=2,      # 布局 1600×1000，输出 3200×2000 像素
    background_color='white',
    colormap='tab10',
    max_words=160, min_font_size=10, max_font_size=220,
    prefer_horizontal=1.0,                 # 全部水平排列，中文最易读
    relative_scaling=0.5,                  # 词频对字号的影响强度
    random_state=7,                        # 固定随机种子，结果可复现
).generate_from_frequencies(freq)          # 直接喂词频字典，比喂文本更可控

plt.figure(figsize=(12, 7.5))
plt.imshow(wc_rect, interpolation='bilinear')
plt.axis('off')
plt.show()
print(f'布局 {wc_rect.width}×{wc_rect.height}，输出 {wc_rect.width * 2}×{wc_rect.height * 2} 像素')
```

```python
# 四、版式二：头像剪影版——把黄院士的肖像变成轮廓，形状本身也承载信息
import numpy as np
from PIL import Image, ImageFilter
from matplotlib.image import imread

photo = imread('huangxuhua.jpg')

# 原图是白底人像照。WordCloud 把"非白色"区域当作可绘制区，
# 但 JPG 压缩会让边界产生毛刺，直接拿来当遮罩会得到锯齿状的轮廓。
# 这里先灰度化 + 阈值二值化，再用高斯模糊加二次阈值把边界磨平。
gray = Image.fromarray(photo[..., :3]).convert('L')
sil = gray.point(lambda v: 0 if v < 245 else 255).filter(ImageFilter.GaussianBlur(3))
sil = sil.point(lambda v: 0 if v < 128 else 255)

# 四周补白边，避免位于边缘的词被裁掉
pad = 46
canvas = Image.new('L', (sil.width + pad * 2, sil.height + pad * 2), 255)
canvas.paste(sil, (pad, pad))
mask = np.array(canvas)

wc_mask = WordCloud(
    font_path=FONT, mask=mask, scale=4,    # 有 mask 时 width/height 失效，靠 scale 放大
    background_color='white', colormap='viridis',
    max_words=160, min_font_size=10, max_font_size=200,
    prefer_horizontal=1.0, relative_scaling=0.5, random_state=7,
).generate_from_frequencies(freq)

plt.figure(figsize=(8, 10.5))
plt.imshow(wc_mask, interpolation='bilinear')
plt.axis('off')
plt.show()
print(f'遮罩 {mask.shape[0]}×{mask.shape[1]}，实际输出 {wc_mask.width * 4}×{wc_mask.height * 4} 像素')
```

```python
# 五、导出 PNG。文字边缘用无损格式更干净，JPG 会在笔画周围产生色斑
for wc, name in [(wc_rect, '黄旭华词云_高清矩形版.png'),
                 (wc_mask, '黄旭华词云_头像剪影版.png')]:
    wc.to_file(name)
    print(f'{name}  {os.path.getsize(name) / 1024:.0f} KB')
```

### 清晰度具体提在哪

| 做法 | 改前 | 改后 | 效果 |
| --- | --- | --- | --- |
| 画布像素 | 400×200，scale=1（约 8 万像素） | 1600×1000，scale=2（约 640 万像素） | 放大不糊，可直接打印 |
| 排版方向 | `prefer_horizontal=0.9`（横竖混排） | `1.0`（全部水平） | 中文竖排极难读，这是最有效的一步 |
| 词数 | 默认 200 词、字号下限 4 | 160 词、字号下限 10 | 小词不再挤成一团 |
| 词表 | 只去 15 个停用词 | 另去单字、纯数字与标点 | 去掉"之一""并且"这类碎片 |
| 输出格式 | JPG | PNG 无损 | 笔画边缘不出现色块 |

### 和上一讲词频结论的呼应

图上 **黄旭华最大、核潜艇次之**，与第 2 讲算出的 53 次、32 次完全对应——**词云本质就是词频的图形化，词频算错，词云也会照错画**。

第 2 讲发现的问题在这里同样成立：原文中"黄旭华"实际出现 59 次，分词只切出 53 次，因为"黄旭华为代表"被切成了"黄旭 / 华为"。**图上你看不出少了 6 次，但统计口径已经偏了。** 这正是可视化最需要注意的地方——它能把错误变得很好看，却不会告诉你哪里错了。

### 还可以怎么改进

1. **让颜色承载信息**：默认 `colormap` 只是随机上色。用 `color_func` 可以让颜色深浅对应词频，读者一眼看出强弱层次。
2. **补充领域停用词**：把"进行""主要""包括""有关"这类高频虚动词加进停用词表，实体词会更突出。
3. **多版本并列对比**：同一份数据出三张不同参数的图并配文字说明，能直观呈现"参数如何影响观感"——正好回应课件第 26 页"还能做哪些工作让词云更好看"。
